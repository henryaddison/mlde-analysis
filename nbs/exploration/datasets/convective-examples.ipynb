```python
%reload_ext autoreload

%autoreload 2

%reload_ext dotenv
%dotenv

import functools
import math
import string

import IPython
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr

from mlde_utils import cp_model_rotated_pole, DatasetMetadata
from mlde_analysis import plot_map
from mlde_analysis.data import open_dataset_split
from mlde_analysis.psd import plot_psd, pysteps_rapsd
```

```python
matplotlib.rcParams['figure.dpi'] = 300
```

```python
dataset_name="demo-ccpm_pr"
split="test"
ensemble_members = ["01", "04"]
var = "target_pr"
```

```python
DS = open_dataset_split(dataset_name, split, ensemble_members).stack(example=["ensemble_member", "time"])
DS
```

```python
raspd_ds = pysteps_rapsd(DS[var].transpose("example", "grid_latitude", "grid_longitude"), pixel_size=8.8)
sorted_by_small_scale_spatial_variability = DS[var].sortby(raspd_ds.isel(freq=-1))
```

```python
ncol = 4
nrow = 2
fig, axs = plt.subplots(nrow, ncol, subplot_kw={"projection": cp_model_rotated_pole}, layout="compressed")

for i in range(nrow):
    for j in range(ncol):
        example_idx = - ((i*ncol) + j+1)
        ax = axs[i][j]
        plot_da = sorted_by_small_scale_spatial_variability.isel(example=example_idx)
        plot_map(plot_da, ax=ax, style="pr")
        ax.set_title(f"{example_idx} {plot_da.example.item()[0]} {plot_da.example.item()[1].strftime()}", fontsize="xx-small")
        
plt.show()
```
