```python
desc = ""
experiment = "Emulator_hist_future"
domain = "SA_domain"
gcm_name = "ACCESSCM2"
split = "train"
target_variables = ["tasmax", "pr"]
framework = "perfect"
static_variables = ["orog"]
predictor_variables = [f"{v}_{p}" for v in ["t", "u", "v", "z", "q"] for p in [500, 700, 850]]
```

```python
import cartopy.feature as cfeature
import IPython
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import os
from pathlib import Path
import xarray as xr

from mlde_utils import platecarree
from mlde_analysis import plot_map
from mlde_analysis.display import pretty_table
```

```python
DATA_PATH = Path(os.getenv("DATA_PATH"))
dataset_name = f"{domain}-{experiment}-{gcm_name}-{framework}"
```

```python
TARGET_DS = xr.open_dataset(DATA_PATH / dataset_name / split / "target" / "pr_tasmax.nc")
TARGET_DS
```

```python
PREDICTOR_DS = xr.open_dataset(DATA_PATH / dataset_name / split / "predictors" / "Variable_fields.nc")
PREDICTOR_DS
```

```python
STATIC_DS = xr.open_dataset(DATA_PATH / dataset_name / split / "predictors" / "Static_fields.nc")
STATIC_DS
```

```python
def plot_variable(da, ax, cmap):
    da.plot(ax=ax, x="lon", y="lat", transform=platecarree, cmap=cmap)

    ax.coastlines(resolution='10m')
    ax.add_feature(cfeature.BORDERS, linestyle=':')
    ax.set_title("")
```

```python
fig = plt.figure(figsize=(4,4))
axd = fig.subplot_mosaic([static_variables], subplot_kw=dict(projection=platecarree))

cmap="terrain"

for var in static_variables:
    ax = axd[var]
    plot_variable(STATIC_DS[var], ax, cmap)

plt.show()
```

```python
fig = plt.figure(figsize=(8,4))
axd = fig.subplot_mosaic([target_variables], subplot_kw=dict(projection=platecarree))

cmap="viridis"

for var in target_variables:
    ax = axd[var]
    plot_variable(TARGET_DS[var].isel(time=0), ax, cmap)

plt.show()
```

```python
fig = plt.figure(figsize=(8.5,11))
axd = fig.subplot_mosaic(np.array(predictor_variables).reshape(5,3), subplot_kw=dict(projection=platecarree))

cmap="viridis"

for var in predictor_variables:
    ax = axd[var]
    plot_variable(PREDICTOR_DS[var].isel(time=0), ax, cmap)

plt.show()
```

```python
fig = plt.figure(figsize=(8.5,11))
axd = fig.subplot_mosaic(np.array(predictor_variables).reshape(5,3), subplot_kw=dict(projection=platecarree))

cmap="viridis"

for var in predictor_variables:
    ax = axd[var]
    plot_variable(PREDICTOR_DS[var].mean("time"), ax, cmap)

plt.show()
```

```python
fig = plt.figure(figsize=(8,4))
axd = fig.subplot_mosaic([target_variables], subplot_kw=dict(projection=platecarree))

cmap="viridis"

for var in target_variables:
    ax = axd[var]
    plot_variable(TARGET_DS[var].mean("time"), ax, cmap)

plt.show()
```
